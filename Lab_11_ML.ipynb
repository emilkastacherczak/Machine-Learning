### KLASYFIKACJA Z WYKORZYSTANIEM SIECI NEURONOWYCH ###

```python
import numpy as np
from keras.datasets import mnist
import keras
from keras import layers
from sklearn.metrics import classification_report, confusion_matrix
```

```python
#Wczytanie i konwersja danych
training_set, test_set = mnist.load_data()

X_train, y_train = training_set
X_test, y_test = test_set

X_train.shape #ilosc, wysokosc i szerokosc obrazu
```

```python
#dodanie wymiaru kanału 
X_train = np.expand_dims(X_train, -1)
X_test = np.expand_dims(X_test, -1)

X_train.shape
```

```python
#przeskalowanie pikseli obrazu - podzielenie każdego przez max mozliwa wartosc
X_train = X_train.astype(np.float64) / 255.0
X_test = X_test.astype(np.float64) / 255.0
```

```python
num_classes = 10 #ilosc klas do klasyfikacji
input_shape = (28, 28, 1) #rozmiar wejscia na siec
batch_size = 128 #rozmiar paczki bedacej jednorazowym wejsciem na siec
epochs = 30 #ilosc epok uczenia
```

```python
model = keras.Sequential(
        [
            keras.Input(shape=input_shape),
            keras.layers.Flatten(),
            layers.Dense(num_classes, activation="softmax")
        ]
    )
```

```python
model.summary()
```

```python
model.compile(loss="sparse_categorical_crossentropy", optimizer="adam", metrics=["accuracy"])
```

```python
model.fit(X_train, y_train, batch_size=batch_size, epochs=epochs, validation_split=0.2)
```

```python
#Ocena modelu
from sklearn.metrics import classification_report, confusion_matrix

y_probab = model.predict(X_test)
y_pred = np.argmax(y_probab, axis=1)

print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
```

#### Uczenie modeli nieliniowych ####

```python
#Sieć MLP powinna składać się z dwóch ukrytych warstw Dense
model = keras.Sequential(
        [
            keras.Input(shape=input_shape),
            keras.layers.Flatten(),
            layers.Dense(64, activation="tanh"),
            layers.Dense(128, activation="tanh"),
            layers.Dense(num_classes, activation="softmax")
        ]
    )
```

```python
model.summary()
```

```python
model.compile(loss="sparse_categorical_crossentropy", optimizer="adam", metrics=["accuracy"])
```

```python
model.fit(X_train, y_train, batch_size=batch_size, epochs=epochs, validation_split=0.2)
```

```python
#Ocena modelu
from sklearn.metrics import classification_report, confusion_matrix

y_probab = model.predict(X_test)
y_pred = np.argmax(y_probab, axis=1)

print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
```

```python
#Sieć konwolucyjna powinna składać się naprzemiennie z warstw konwolucyjnych oraz maxpooling
model = keras.Sequential(
        [
            keras.Input(shape=input_shape),
            layers.Conv2D(32, kernel_size=(3, 3), activation="relu"),
            layers.MaxPooling2D(pool_size=(2, 2)),
            layers.Conv2D(64, kernel_size=(3, 3), activation="relu"),
            layers.MaxPooling2D(pool_size=(2, 2)),
            layers.Flatten(),
            layers.Dropout(0.5),
            layers.Dense(num_classes, activation="softmax"),
        ]
    )
```

```python
model.summary()
```

```python
model.compile(loss="sparse_categorical_crossentropy", optimizer="adam", metrics=["accuracy"])
```

```python
model.fit(X_train, y_train, batch_size=batch_size, epochs=epochs, validation_split=0.2)
```

```python
#Ocena modelu
from sklearn.metrics import classification_report, confusion_matrix

y_probab = model.predict(X_test)
y_pred = np.argmax(y_probab, axis=1)

print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
```
